## 📄 Article 1 : Munirathinam & Yingling (1994)

**Référence complète** :
- Auteurs : Munirathinam, M. & Yingling, J.C.
- Année : 1994
- Titre : *A review of computer-based truck dispatching strategies for surface mining operations*
- Source : International Journal of Surface Mining, Reclamation and Environment
- Fichier PDF : `A review of computer-based truck dispatching strategies for surface mining operations.pdf`

**⭐ PRIORITÉ** : Article clé pour comprendre les heuristiques de base (tes futures baselines)

---

### ❓ Question 1 : Quel est le problème principal que cet article aborde ?

**Réponse** :

*L'article traite des systèmes informatisés de répartition des camions dans les mines à ciel ouvert, qui visent à améliorer l'utilisation des équipements et la productivité tout en respectant des contraintes opérationnelles comme le contrôle de la teneur en minerai et les ratios de décapage. L'objectif est de classifier et d'analyser les différentes stratégies existantes.*

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils pour l'analyser ?

**Réponse** :

*Les auteurs classent les stratégies de répartition en plusieurs catégories : les systèmes heuristiques basés sur des règles, les systèmes basés sur des plans ("plan-based"), et l'approche d'affectation contrainte de Hauck. Ils examinent en détail les formulations mathématiques sous-jacentes de chaque approche et identifient leurs forces et faiblesses respectives.*

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

*Les systèmes heuristiques basés sur des règles sont simples à implémenter mais présentent des limites (décisions myopes, difficulté à gérer plusieurs objectifs). Les systèmes basés sur des plans corrigent ces faiblesses en définissant un plan opérationnel optimal, tandis que l'approche d'affectation contrainte de Hauck offre une meilleure prise en compte collective des camions. Les auteurs soulignent le manque d'études de simulation comparatives entre ces différentes approches.*

---

### 🎯 Informations Spécifiques pour Baseline

- **Algorithmes/Heuristiques mentionnés** :
    - Fixed Truck Assignment (FTA)
    - Règles heuristiques simples (Minimize Waiting Time, Minimize Queue Length, etc.)
    - Systèmes basés sur des plans (DISPATCH, LP-based)
    - Approche d'affectation contrainte de Hauck (DP + LP)
- **Métriques de performance utilisées** :
    - Productivité (tonnes/heure)
    - Temps de cycle
    - Utilisation de la flotte
    - Respect des contraintes de teneur et ratios de décapage
    - Temps d'attente aux pelles
- **Implémentation possible** : Facile pour les heuristiques simples (règles), Moyen pour les systèmes plan-based
- **Code/Pseudo-code disponible** : Non (formulations mathématiques décrites mais pas de pseudo-code explicite)

---

### ❓ Question 1 : Quel est le problème principal ?

*L'article traite du dispatching dynamique des camions dans les mines à ciel ouvert, où le transport représente environ 50% des coûts opérationnels. Le défi est de concevoir une méthode applicable en conditions réelles, capable de gérer les incertitudes (pannes, retards, réaffectations) tout en maximisant la productivité.*

La méthode est validée sur des données industrielles réelles d'une mine de fer et intégrée au logiciel SmartMine®.*

### ❓ Question 3 : Conclusion principale ?

*L'approche permet une meilleure utilisation des camions, une augmentation de la production et une réduction des coûts opérationnels par rapport aux performances réellement observées sur site. La combinaison LP + simulation + heuristiques dynamiques s'avère robuste et industriellement viable.*

## 📄 Article 3 : Souza et al. (2010)

**Référence complète** :
- Auteurs : Souza, M.J.F. et al.
- Année : 2010
- Titre : *A hybrid heuristic algorithm for the open-pit-mining operational planning problem*
- Fichier PDF : `A hybrid heuristic algorithm for the open-pit-mining operational planning problem.pdf`

---

### ❓ Question 1 : Quel est le problème principal ?

*L'article traite du problème de planification opérationnelle des mines à ciel ouvert (OPMOP), un problème NP-difficile qui consiste à déterminer les taux d'extraction, l'affectation des pelles et le nombre de trajets des camions tout en respectant les objectifs de production, les contraintes de qualité du minerai (teneurs en Fe, SiO₂, P) et en minimisant le nombre de camions utilisés.*

### ❓ Question 2 : Méthodologie ?

*Les auteurs développent un modèle de programmation linéaire en nombres entiers mixtes (MIP) basé sur la programmation par objectifs, puis proposent un algorithme heuristique hybride combinant GRASP (construction de solutions initiales diversifiées) et VNS (amélioration locale via 8 types de mouvements de voisinage). L'approche est validée sur des données réelles de mines.*

### ❓ Question 3 : Conclusion principale ?

*L'algorithme hybride GRASP-VNS atteint des solutions avec un écart à l'optimum inférieur à 1% dans la majorité des cas, avec des temps de calcul très réduits compatibles avec un usage opérationnel. Cette approche démontre qu'il est possible d'obtenir des solutions quasi-optimales pour des problèmes industriels complexes en temps réel.

*

---

### 🎯 Informations Spécifiques pour Baseline

- **Algorithmes/Heuristiques** :
    - Modèle MIP (programmation par objectifs)
    - GRASP (Greedy Randomized Adaptive Search Procedure)
    - VNS (Variable Neighborhood Search) avec 8 voisinages
    - Allocation dynamique des camions
- **Métriques** :
    - Écart à l'optimum (%)
    - Temps de calcul
    - Nombre de camions utilisés
    - Respect des objectifs de production
    - Qualité du mélange (teneurs)
- **Implémentation** : Moyen-Difficile (MIP + métaheuristiques combinées)

- Titre : *Real-time dispatching modelling for trucks with different capacities in open pit mines*
- Fichier PDF : `Real Time Dispatching Modelling For Trucks With Different Capacities In Open Pit Mines.pdf`

---

### ❓ Question 1 : Quel est le problème principal ?

*L'article aborde le défi du dispatching en temps réel pour des flottes hétérogènes de camions (capacités différentes) dans les mines à ciel ouvert. Les systèmes existants étaient soit statiques, soit conçus pour des flottes homogènes, ce qui limitait leur efficacité face à la réalité opérationnelle des mines modernes utilisant différents types de camions.*

### ❓ Question 2 : Méthodologie ?

*Les auteurs proposent un modèle hybride combinant la théorie des réseaux (Flow Network) pour optimiser les itinéraires et calculer les temps de trajet, puis une programmation linéaire en nombres entiers binaires (BIP) pour l'allocation dynamique des camions. Le modèle est recalculé à chaque étape de dispatching, le rendant adaptatif aux conditions réelles.*

### ❓ Question 3 : Conclusion principale ?

*Le modèle permet de réduire les temps d'attente et d'améliorer le taux d'utilisation des équipements en intégrant explicitement les différences de capacité entre camions. C'est le premier modèle temps réel à gérer efficacement les flottes hétérogènes, constituant un pont entre les approches classiques et les méthodes modernes (IA, RL).*

---

## 📄 Article 5 : Ta et al. (2013)

**Référence complète** :
- Auteurs : Ta, C.H., Ingolfsson, A., Doucette, J.
- Année : 2013
- Titre : *A linear model for surface mining haul truck allocation incorporating shovel idle probabilities*
- Source : European Journal of Operational Research, 231, 770-778
- Fichier PDF : `A linear model for surface mining haul truck allocation incorporating shovel idle probabilities.pdf`

---

### ❓ Question 1 : Quel est le problème principal ?

*L'article traite de l'allocation optimale d'un nombre minimal de camions à un ensemble de pelles, tout en respectant les contraintes de débit et de qualité du minerai. Le défi est d'intégrer la probabilité d'inactivité des pelles (phénomène stochastique issu des files d'attente) dans un modèle d'optimisation linéaire tractable.*

### ❓ Question 2 : Méthodologie ?

*Les auteurs proposent un modèle linéaire en nombres entiers mixtes (MILP) qui linéarise la probabilité d'inactivité des pelles à partir de la théorie des files d'attente à source finie. Cette approximation permet d'exprimer le débit de la pelle comme une fonction linéaire du nombre de camions alloués, rendant le problème optimisable efficacement.*

### ❓ Question 3 : Conclusion principale ?

*La linéarisation de la probabilité d'inactivité est suffisamment précise et le modèle reste facilement solvable à l'échelle industrielle. Les allocations obtenues sont plus cohérentes que celles basées sur l'expérience seule. Le modèle sert d'allocation initiale de référence pour les systèmes de dispatching en temps réel.*

---

### 🎯 Informations Spécifiques pour Baseline

- **Algorithmes/Heuristiques** :
    - MILP (Programmation linéaire en nombres entiers mixtes)
    - Approximation linéaire des probabilités d'inactivité
    - Théorie des files d'attente à source finie
- **Métriques** :
    - Nombre total de camions (à minimiser)
    - Débit par pelle (tonnes/heure)
    - Probabilité d'inactivité des pelles
    - Contraintes de teneur (qualité minerai)
    - Production par période
- **Implémentation** : Moyen (MILP standard, solvable avec CPLEX/Gurobi)
- **Code disponible** : Non

## 📄 Article 6 : Wang et al. (2022)

**Référence complète** :
- Auteurs : Wang, Z. et al.
- Année : 2022
- Titre : *Truck Dispatching Optimization Model and Algorithm Based on 0-1 Decision Variables*
- Source : Mathematical Problems in Engineering
- Fichier PDF : `Truck Dispatching Optimization Model and Algorithm Based on 0-1 Decision Variables.pdf`

---

### ❓ Question 1 : Quel est le problème principal ?

*L'article traite du dispatching des camions dans les mines à ciel ouvert avec l'objectif de maximiser le volume total de chargement et de déchargement sur une période de travail. Le défi est de formaliser les décisions d'affectation camion-trajet de manière exploitable industriellement, en gérant les contraintes opérationnelles réelles.*

### ❓ Question 2 : Méthodologie ?

*Les auteurs proposent un modèle de programmation linéaire en nombres entiers (ILP) utilisant des variables de décision binaires (0-1) pour représenter si un camion effectue ou non un trajet donné. Le modèle intègre des contraintes de temps de travail, capacités, flux de trafic et cohérence logique. La résolution est effectuée avec le solveur LINGO.*

### ❓ Question 3 : Conclusion principale ?

*Le modèle atteint un volume total transporté de 10 950,6 m³ avec 384 opérations de chargement/déchargement sur le cas d'étude. Les résultats sont suffisamment précis pour guider directement la production. L'approche par variables binaires produit des solutions concrètes et exploitables, bien que limitée par son caractère statique.*

---

## 📄 Article 1 : Munirathinam & Yingling (1994)

**Référence complète** :
- Auteurs : Munirathinam, M. & Yingling, J.C.
- Année : 1994
- Titre : *A review of computer-based truck dispatching strategies for surface mining operations*
- Source : International Journal of Surface Mining, Reclamation and Environment
- Fichier PDF : `A review of computer-based truck dispatching strategies for surface mining operations.pdf`

**⭐ PRIORITÉ** : Article clé pour comprendre les heuristiques de base (tes futures baselines)

---

### ❓ Question 1 : Quel est le problème principal que cet article aborde ?

**Réponse** :

*L'article traite des systèmes informatisés de répartition des camions dans les mines à ciel ouvert, qui visent à améliorer l'utilisation des équipements et la productivité tout en respectant des contraintes opérationnelles comme le contrôle de la teneur en minerai et les ratios de décapage. L'objectif est de classifier et d'analyser les différentes stratégies existantes.*

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils pour l'analyser ?

**Réponse** :

*Les auteurs classent les stratégies de répartition en plusieurs catégories : les systèmes heuristiques basés sur des règles, les systèmes basés sur des plans ("plan-based"), et l'approche d'affectation contrainte de Hauck. Ils examinent en détail les formulations mathématiques sous-jacentes de chaque approche et identifient leurs forces et faiblesses respectives.*

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

*Les systèmes heuristiques basés sur des règles sont simples à implémenter mais présentent des limites (décisions myopes, difficulté à gérer plusieurs objectifs). Les systèmes basés sur des plans corrigent ces faiblesses en définissant un plan opérationnel optimal, tandis que l'approche d'affectation contrainte de Hauck offre une meilleure prise en compte collective des camions. Les auteurs soulignent le manque d'études de simulation comparatives entre ces différentes approches.*

---

### 🎯 Informations Spécifiques pour Baseline

- **Algorithmes/Heuristiques mentionnés** :
    - Fixed Truck Assignment (FTA)
    - Règles heuristiques simples (Minimize Waiting Time, Minimize Queue Length, etc.)
    - Systèmes basés sur des plans (DISPATCH, LP-based)
    - Approche d'affectation contrainte de Hauck (DP + LP)
- **Métriques de performance utilisées** :
    - Productivité (tonnes/heure)
    - Temps de cycle
    - Utilisation de la flotte
    - Respect des contraintes de teneur et ratios de décapage
    - Temps d'attente aux pelles

- **Code/Pseudo-code disponible** : Non (formulations mathématiques décrites mais pas de pseudo-code explicite)

## 📄 Article 1 : Nazari et al. (2018) ⭐⭐⭐

**Référence complète** :
- Auteurs : Nazari, M., Oroojlooy, A., Snyder, L., Takác, M.
- Année : 2018
- Titre : *Reinforcement Learning for Solving the Vehicle Routing Problem*
- Source : NeurIPS
- Fichier PDF : `Reinforcement Learning for Solving the Vehicle Routing Problem.pdf`

**🔥 ARTICLE CLÉ** : C'est LE papier qui va inspirer ta méthodologie. À lire EN PROFONDEUR.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

*Résoudre le Vehicle Routing Problem (VRP) en utilisant une approche d’apprentissage par renforcement (RL), sans recourir aux heuristiques classiques ou solveurs experts. L’objectif est d’apprendre une politique de construction de tournée optimale pour une flotte de véhicules, sous contraintes de capacité et de demande.*

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

*Ils modélisent le VRP comme un processus de décision markovien (MDP), où l’état inclut la position actuelle, la capacité restante et les demandes des clients. Ils utilisent un réseau encodeur-décodeur basé sur LSTM avec un mécanisme d’attention, et entraînent le modèle par policy gradient (REINFORCE). Les actions invalides sont masquées dynamiquement, et l’état est mis à jour à chaque étape.*

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

*Un agent RL peut apprendre à construire des tournées pour le VRP sans heuristique humaine, avec des performances proches des heuristiques classiques sur des instances de taille moyenne, et une bonne généralisation. Cependant, il reste inférieur aux meilleurs solveurs spécialisés et nécessite un entraînement préalable.*

---

### 🧠 Détails Techniques pour Implémentation RL

#### 1️⃣ Définition du MDP

**État (State)** :
- Représentation : *Coordonnées des clients, demande de chaque client, capacité restante du véhicule, position actuelle, demandes restantes.*
- Dimension : *Variable selon le nombre de clients (ex : 10 à 100).*
- Type : Hybride (coordonnées continues, demandes discrètes).

**Action (Action)** :
- Espace d'action : *Sélectionner le prochain client à visiter ou retourner au dépôt.*
- Dimension : *Discrète, taille = nombre de clients + 1 (dépôt).*
- Type : Discret.

**Récompense (Reward)** :
- Formule : *R = -distance_totale parcourue (minimiser la distance).*
- Pénalités : *Masquage des actions invalides (ex : client dont la demande > capacité restante).*

**Transition** :
- Déterministe (choix du client → mise à jour état, capacité, position).

---

#### 2️⃣ Architecture du Réseau de Neurones

- **Type de réseau** : *Encodeur-décodeur LSTM avec attention (Pointer Network modifié).*
- **Nombre de couches** : *1-2 couches LSTM (encodeur/décodeur).*
- **Fonction d'activation** : *ReLU pour embeddings, Tanh pour LSTM.*
- **Mécanisme attention** : Oui, permet de pointer dynamiquement les clients à visiter.

---

#### 3️⃣ Algorithme RL Utilisé

- **Nom** : *Policy Gradient (REINFORCE)*
- **On-policy / Off-policy** : On-policy
- **Hyperparamètres clés** : *Learning rate (ex : 1e-4), discount factor γ (ex : 0.99), baseline critique pour variance.*

---

#### 4️⃣ Entraînement

- **Nombre d'épisodes** : *Plusieurs centaines de milliers (ex : 500k+)*
- **Durée d'entraînement** : *Plusieurs heures à jours (selon hardware)*
- **Hardware utilisé** : *GPU (NVIDIA)*
- **Curriculum learning** : Non (mais possible dans extensions ultérieures)

---

#### 5️⃣ Résultats et Métriques

- **Métriques utilisées** : *Distance totale parcourue, temps d’inférence, gap avec optimal.*
- **Baselines comparées** : *OR-Tools, heuristiques classiques (Clarke-Wright, Sweep).*
- **Performance** : *Proche des heuristiques classiques, généralisation correcte, mais inférieur aux solveurs exacts.*

---

## 📄 Article 2 : Liu & Chai (2019)

**Référence complète** :
- Auteurs : Liu, X. & Chai, X.
- Année : 2019
- Titre : *Optimizing Open-Pit Truck Route Based on Minimization of Time-Varying Transport Energy Consumption*
- Source : Mathematical Problems in Engineering
- Fichier PDF : `Optimizing Open-Pit Truck Route Based on Minimization of Time-Varying Transport Energy Consumption.pdf`

**🎯 FOCUS** : Fonction de récompense basée sur la consommation d'énergie/carburant.

---

### ❓ Question 1 : Quel est le problème principal ?

*Optimiser les itinéraires des camions dans une mine à ciel ouvert en minimisant la consommation d’énergie de transport qui varie dans le temps, en tenant compte de la dégradation dynamique des routes (résistance au roulement variable), plutôt que de se limiter à la distance ou au temps de trajet.*

### ❓ Question 2 : Méthodologie ?

*Les auteurs modélisent le réseau minier comme un graphe orienté et formulent le problème comme un programme mixte en nombres entiers (MIP), où la consommation énergétique sur chaque segment de route est une variable aléatoire dépendant du temps.
Ils estiment la résistance au roulement via des méthodes statistiques (pattern recognition, trend surface estimation) et résolvent le problème à l’aide d’un algorithme génétique amélioré pour gérer la complexité et la dynamique temporelle.*

### ❓ Question 3 : Conclusion ?

*Leur approche dynamique permet d’obtenir des itinéraires plus économes en énergie que les méthodes classiques statiques. L’algorithme génétique amélioré converge rapidement et donne des solutions stables, mais la méthode dépend fortement de la qualité des données historiques et a été validée sur un seul site industriel.*

---

### 🧮 Modèle de Consommation d'Énergie

**Formule principale** :

```
Eij(t) = Dij⋅fij(t)
```
où :

- E*ij*(t) = consommation d’énergie sur l’arc (i,j) au temps t
- D*ij* = distance du segment (i,j)
- f*ij*(t) = coefficient de résistance au roulement variable dans le temps

**Variables prises en compte** :
- Vitesse : *Oui (influence indirecte via fij(t))*
- Charge du camion : *Oui (dans le calcul de la résistance)*
- Pente de la route : *Oui (intégrée dans fij(t))*
- État de la route : *Oui (cœur du modèle, dynamique)*
- Conditions météo : *Non explicitement, mais possible via f*ij*(t) si les données sont disponibles*

**Paramètres du modèle** :
- f*ij*(t) : estimé statistiquement, dépend de l’usure, maintenance, etc.
- D : *constante de pondération pour la fitness*
---

### 🔗 Intégration dans ma Fonction de Récompense

**Récompense proposée** :

```
R = -α·E(fuel) - β·Time - γ·WaitTime - δ·Wear
```

Où `E(fuel)` sera calculé selon la formule de Liu & Chai (2019), en intégrant la résistance au roulement dynamique.

**Poids à ajuster** : α, β, γ, δ selon l’importance relative de chaque critère.

## 📄 Article 3 : Zhang et al. (2020)

**Référence complète** :
- Auteurs : Zhang, C. et al.
- Année : 2020
- Titre : *Learning to Dispatch for Job Shop Scheduling via Deep Reinforcement Learning*
- Source : NeurIPS
- Fichier PDF : `Learning to Dispatch for Job Shop Scheduling via Deep Reinforcement Learning.pdf`

**🚀 AVANCÉ** : Utilisation de Graph Neural Networks (GNN) — À lire si temps disponible.

---

### ❓ Question 1 : Quel est le problème principal ?

*Remplacer les règles de dispatching heuristiques (PDR) pour le Job Shop Scheduling Problem (JSSP) par une politique apprise automatiquement via Deep Reinforcement Learning, afin de minimiser le makespan sans expertise humaine.*

### ❓ Question 2 : Quelle méthodologie ?

*Le JSSP est formulé comme un MDP. L’état est représenté par un graphe disjonctif, encodé par un Graph Neural Network (GIN). L’agent apprend une politique de dispatching end-to-end, à partir de features brutes, sans règles expertes.*

### ❓ Question 3 : Conclusion principale ?

*L’agent DRL+GNN surpasse les meilleures règles de dispatching classiques, généralise à des instances plus grandes, et fonctionne sans ingénierie experte, mais au prix d’une interprétabilité réduite.*

---

### 🧠 Architecture GNN

**Pourquoi un graphe ?**
- *Le planning du JSSP (et, par analogie, le réseau minier) est naturellement structuré comme un graphe : les opérations (ou points du réseau) sont des nœuds, les contraintes/précédences ou routes sont des arêtes. Cela permet de capturer la structure globale et les dépendances.*

**Type de GNN utilisé** :
- Graph Isomorphism Network (GIN) : *un GNN puissant pour extraire des embeddings structurés et différenciables.*

**Représentation des nœuds** :
- Pour le JSSP : *Chaque opération (job, machine, temps de traitement, état d’avancement).*
- *Pour la logistique minière : points de chargement, déchargement, intersections, zones tampons, occupation, temps d’attente, priorité.*

**Représentation des arêtes** :
- Pour le JSSP : *Contraintes de précédence, ordre sur machine.*
- Pour la logistique minière : *Routes (distance, état, pente, trafic, consommation estimée).*

---

### 🔗 Transférabilité

**Mon graphe minier pourrait avoir** :
- **Nœuds** : Points de chargement, déchargement, intersections, zones tampons
- **Arêtes** : Routes avec attributs (distance, état, pente, trafic)
- **Features nœuds** : Occupation, temps d’attente, priorité, type de point
- **Features arêtes** : État de la route, consommation estimée, capacité, vitesse limite

**Complexité d'implémentation** : Élevée — nécessite des bibliothèques spécialisées (PyTorch Geometric, DGL), gestion de graphes dynamiques, adaptation des features.

## 📄 Article 4 : Optimization-Based Dispatching Policies

**Référence** :
- Auteurs : Smith, J.C., Linderoth, J.T., Luedtke, J.R.
- Année : 2021
- Titre : *Optimization-Based Dispatching Policies for Open-Pit Mining*
- Source : Springer
- Fichier : `Optimization-Based Dispatching Policies for Open-Pit Mining.pdf`

---

### ❓ Question 1 : Quel est le problème principal ?

*Optimiser l’affectation dynamique des camions dans une mine à ciel ouvert pour maximiser la production et l’utilisation des ressources, tout en minimisant les temps d’attente et les congestions, en allant au-delà des politiques industrielles classiques basées sur des moyennes de flux.*

### ❓ Question 2 : Quelle méthodologie ?

*Deux approches sont développées et comparées* :

*Modèle non linéaire basé sur les flux moyens : calcule des taux de flux optimaux entre zones, intégrant explicitement les files d’attente.
Modèle MIP (Mixed-Integer Programming) discrétisé dans le temps : modélise chaque décision d’affectation de camion à chaque pas de temps, résolu par heuristique pour garantir la rapidité.
Les deux politiques sont évaluées dans un simulateur à événements discrets représentant fidèlement la dynamique d’une mine réelle.*

### ❓ Question 3 : Conclusion principale ?

*La politique basée sur le MIP, même résolue de façon approchée, surpasse systématiquement la politique basée sur les flux moyens, en termes de production, d’utilisation des camions et de réduction des temps d’attente, sur une large gamme de scénarios opérationnels.*

---

### 🧠 Détails Techniques

- Modèle flux moyens :
    - Non linéaire, intègre files d’attente (queueing theory)
    - Fournit des cibles de flux entre zones (ex : pelles → concasseur)
    - Approche statique, adaptée à la planification globale

- Modèle MIP :
    - Variables binaires pour l’affectation de chaque camion à chaque tâche à chaque pas de temps
    - Contraintes : disponibilité, capacité, séquences, etc.
    - Trop complexe pour résolution exacte en temps réel → heuristique développée pour obtenir des solutions rapides et de qualité

- Simulation :
    - Événements discrets, dynamique stochastique
    - Mesures : production, utilisation, temps d’attente, files, efficacité

### 🔗 Transférabilité à mon projet

- Ce que je peux réutiliser :
     1. L’idée de comparer une politique “statique” (flux moyens) à une politique “dynamique” (MIP/heuristique ou RL)
     2. L’intégration d’un simulateur réaliste pour évaluer les politiques
     3. Les métriques d’évaluation (production, utilisation, files, etc.)

- Ce que je dois adapter :
    1. Remplacer la politique MIP par une politique RL (ou hybride)
    2. Adapter le simulateur à la logistique minière spécifique (routes, états, consommation, etc.)
    3. Intégrer la stochasticité (dégradation, météo, etc.)

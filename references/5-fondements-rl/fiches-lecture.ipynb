## 📄 Article 1 : Sutton & Barto (2018) ⭐⭐⭐

**Référence complète** :
- Auteurs : Sutton, R. S. ; Barto, A. G.
- Année : 2018 (2e édition)
- Titre : *Reinforcement Learning: An Introduction*
- Source : MIT Press, Cambridge (MA) — version en ligne gratuite des auteurs (incompleteideas.net)
- Fichier PDF : `Reinforcement Learning An Introduction (2nd ed).pdf` (présent, version en ligne de 2020)

**⭐ Intérêt** : Manuel de référence ; base de nos définitions (MDP, valeur, Q-learning, SARSA, gradient de politique).

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Présenter de façon unifiée l'apprentissage par renforcement : apprendre à agir par interaction pour maximiser une récompense cumulée.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Partie I, méthodes tabulaires : bandits, MDP finis (chap. 3), programmation dynamique, Monte-Carlo, différences temporelles (chap. 6 : **SARSA §6.4**, **Q-learning §6.5**, Expected SARSA §6.6), méthodes à n pas, planification. Partie II, approximation : prédiction et contrôle avec approximation (chap. 9-10, dont la récompense moyenne §10.3), méthodes hors politique et **« triade mortelle » §11.3** (approximation + amorçage + apprentissage hors politique), traces d'éligibilité, **gradient de politique et acteur-critique (chap. 13)**. Partie III : psychologie, neurosciences, applications, perspectives (chap. 17, dont les options).

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Cadre commun des méthodes de valeur et de politique ; compromis exploration-exploitation ; rôle de l'actualisation γ.

---

### 🔗 Pour mon article

- Référence des définitions de base (§5.4) et des pseudocodes Q-learning et SARSA.
- Les deux méthodes tabulaires que nous utilisons y sont présentées côte à côte (SARSA sur politique, Q-learning hors politique), ce qui aide à expliquer leurs comportements différents.

## 📄 Article 2 : Watkins & Dayan (1992) ⭐⭐⭐

**Référence complète** :
- Auteurs : Watkins, C. J. C. H. ; Dayan, P.
- Année : 1992
- Titre : *Q-learning* (note technique)
- Source : Machine Learning, 8(3-4), 279-292
- DOI / Lien : https://doi.org/10.1007/BF00992698
- Fichier PDF : `Q-learning.pdf` (présent, copie de l'auteur)

**⭐ Intérêt** : Référence du Q-learning et de sa convergence.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Démontrer en détail la convergence du Q-learning, présenté par Watkins en 1989.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Le Q-learning est vu comme une programmation dynamique asynchrone et incrémentale, sans modèle ; preuve de convergence fondée sur un processus de « rejeu des actions ».

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Le Q-learning converge vers les valeurs d'action optimales avec probabilité 1 si **toutes les actions sont essayées une infinité de fois dans tous les états**, si les valeurs sont **représentées en table** et si les pas d'apprentissage décroissent convenablement. Extensions esquissées : cas absorbant non actualisé, mises à jour multiples.

---

### 🔗 Pour mon article

- Nos conditions s'en écartent : α constant, exploration qui s'arrête à ε = 0,01, état agrégé (6 variables) qui n'est pas markovien. Cela aide à expliquer la mauvaise convergence de nos agents tabulaires (§6.3, limites).

## 📄 Article 3 : Rummery & Niranjan (1994) ⭐⭐

**Référence complète** :
- Auteurs : Rummery, G. A. ; Niranjan, M.
- Année : 1994
- Titre : *On-Line Q-Learning Using Connectionist Systems*
- Source : Technical Report CUED/F-INFENG/TR 166, Cambridge University Engineering Department
- Fichier PDF : `On-Line Q-Learning Using Connectionist Systems.pdf` (présent)

**⭐ Intérêt** : Référence d'origine de l'algorithme SARSA.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Étendre le Q-learning à des espaces d'états continus de grande dimension avec des réseaux de neurones, et apprendre en ligne pendant les essais plutôt qu'en rejouant l'essai à la fin.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Plusieurs règles fondées sur le Q-learning et TD(λ), dont une nouvelle, le **Modified Connectionist Q-Learning (MCQ-L)**, qui met à jour vers la valeur de l'action réellement choisie à l'état suivant : c'est l'algorithme appelé ensuite SARSA. Comparaison sur la navigation d'un robot mobile simulé parmi des obstacles.

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

L'apprentissage en ligne est moins sensible aux paramètres que le rejeu ; MCQ-L et Q(λ) sont plus robustes que le Q-learning standard.

---

### 🔗 Pour mon article

- Le nom « SARSA » n'apparaît pas dans ce rapport (proposé ensuite par Sutton). Au besoin, écrire « SARSA, introduit sous le nom de Modified Connectionist Q-Learning ».

## 📄 Article 5 : Schulman et al. (2017) — PPO ⭐⭐⭐

**Référence complète** :

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

## 📄 Article 8 : Bradtke & Duff (1995) ⭐⭐⭐

**Référence complète** :
- Auteurs : Bradtke, S. J. ; Duff, M. O.
- Année : 1995 (conférence NIPS 1994)
- Titre : *Reinforcement Learning Methods for Continuous-Time Markov Decision Problems*
- Source : Advances in Neural Information Processing Systems 7, MIT Press, p. 393-400
- DOI / Lien : https://papers.nips.cc/paper_files/paper/1994/hash/07871915a8107172b3b5dc15a6574ad3-Abstract.html
- Fichier PDF : `Reinforcement Learning Methods for Continuous-Time Markov Decision Problems.pdf` (présent)

**⭐ Intérêt** : Fondement direct de notre actualisation par minute.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Les algorithmes de RL (TD(λ), Q-learning, RTDP) supposent un temps discret, alors que beaucoup de systèmes (files d'attente, routage) évoluent en temps continu, avec des durées de transition aléatoires.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Processus de décision semi-markovien (SMDP) ; équation de Bellman où l'actualisation **e^(−βτ) dépend de la durée τ** de la transition ; versions SMDP de TD(0), du Q-learning (cible (1 − e^(−βτ))/β · r + e^(−βτ) max Q) et de l'Adaptive RTDP. Si toutes les durées valent 1, on retrouve le MDP discret.

Expérience : routage d'arrivées vers deux files à serveurs exponentiels ; le Q-learning SMDP avec réseau de neurones approche la politique optimale à seuil.

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Les méthodes de RL s'étendent naturellement aux SMDP en remplaçant γ par une actualisation qui dépend de la durée observée.

---

### 🔗 Pour mon article

- Notre γ_min^Δt joue le rôle de e^(−βτ) (β = −ln γ_min) ; Δt = durée réelle de la décision.
- Notre test d'équivalence (toutes les durées égales à 1,89 min → apprentissage précédent) correspond à la remarque des auteurs : durées identiques → MDP discret.
- Problème d'application proche du nôtre (envoyer un client vers une file).

## 📄 Article 9 : Sutton, Precup & Singh (1999) ⭐⭐⭐

**Référence complète** :
- Auteurs : Sutton, R. S. ; Precup, D. ; Singh, S.
- Année : 1999
- Titre : *Between MDPs and semi-MDPs: A framework for temporal abstraction in reinforcement learning*
- Source : Artificial Intelligence, 112(1-2), 181-211
- DOI / Lien : https://doi.org/10.1016/S0004-3702(99)00052-1
- Fichier PDF : `Between MDPs and semi-MDPs.pdf` (présent, copie de cours UMass)

**⭐ Intérêt** : Deuxième fondement de la formulation semi-markovienne.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Représenter et apprendre des actions qui durent plusieurs pas de temps (abstraction temporelle) dans le cadre des MDP.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Les **options** : politiques en boucle fermée avec conditions d'initiation et de terminaison. Théorème 1 : un MDP avec des options exécutées jusqu'à leur terme est un SMDP. SMDP Q-learning : Q(s, o) ← Q(s, o) + α[r + γ^k max Q(s', o') − Q(s, o)], où k est le nombre de pas écoulés et r la récompense actualisée cumulée. Apports propres : interruption d'options, apprentissage intra-option, sous-buts ; exemples sur un monde de pièces.

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Options et actions primitives s'utilisent indifféremment en planification et en apprentissage.

---

### 🔗 Pour mon article

- Justifie la forme γ^k quand une décision dure k unités de temps ; notre γ_min^Δt en est la version continue (Δt en minutes).
- « Aller à telle pelle » est une option dont la durée est aléatoire.

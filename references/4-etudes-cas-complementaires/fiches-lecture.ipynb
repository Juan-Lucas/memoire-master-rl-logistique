## 📄 Article 1 : Ozdemir & Kumral (2019)

**Référence** :
- Titre : *Simulation based optimization of truck shovel material handling systems in multi pit surface mines*
- Fichier : `Simulation based optimization of truck shovel material handling systems in multi pit surface mines.pdf`

---

### 📊 Résultats Clés

**Métriques utilisées** :
- Productivité totale (t/shift)
- Utilisation flotte camions (%)
- Utilisation pelles (%)
- Temps d’attente camions (min)
- Temps d’attente pelles (min)
- Match Factor (MF)
- Production supplémentaire (t/shift)*

**Chiffres importants** :
- Amélioration de +9,4 % de la production totale par poste (environ 6 000 tonnes/shift)
- Utilisation des équipements accrue
- Réduction des temps d’attente
- Temps de calcul : non précisé, mais compatible avec une planification opérationnelle (pas temps réel)

---

### ⚠️ Limites Identifiées

**Ce qui ne fonctionne pas bien** :
- Validation sur un seul site minier réel
- Dépendance à la qualité des données opérationnelles
- Complexité d’implémentation en temps réel (simulation + optimisation)
- Hypothèses sur la distribution des temps de cycle

**Hypothèses simplificatrices** :
- Les distributions de temps de cycle sont supposées connues et stables
- Pas de prise en compte explicite de la dégradation des routes ou de la météo

---

### 💡 Ce que je peux en tirer pour mon projet

**Insights pratiques** :
1. Le "match factor" (MF) est un KPI central pour équilibrer flotte et pelles, à suivre dans toute simulation minière.
2. L’approche hybride simulation + optimisation mathématique est très pertinente pour intégrer incertitude et contraintes réelles.
3. La réaffectation dynamique inter-fosses améliore significativement la productivité.

**Métriques à réutiliser** :
- Productivité (t/shift ou t/h)
- Utilisation flotte camions et pelles (%)
- Temps d’attente camions/pelles (min)
- Match Factor (MF)
- Production supplémentaire vs baseline

## 📄 Article 2 : Abolghasemian et al. (2020)

**Référence** :
- Titre : *A Two-Phase Simulation-Based Optimization of Hauling System in Open-Pit Mine*
- Fichier : `A Two-Phase Simulation-Based Optimization of Hauling System in Open-Pit Mine.pdf`

---

### 📊 Résultats Clés

**Métriques utilisées** :
- Production totale (t/shift)
- Nombre optimal de pelles
- Utilisation du concasseur (%)
- Coût de transport ($/t)
- Absolute Relative Error (ARE) pour validation du méta-modèle

**Chiffres importants** :
- Augmentation de la production de 21 % (~10 000 tonnes supplémentaires)
- Amélioration directe du revenu (prix du cuivre ≈ 6 000 $/t)
- Allocation optimale des ressources (pelles, camions)
- Réduction des coûts opérationnels relatifs

---

### ⚠️ Limites Identifiées

**Ce qui ne fonctionne pas bien** :
- Étude basée sur un seul cas réel (mine de cuivre iranienne)
- Dépendance à la qualité des données de simulation

**Hypothèses simplificatrices** :
- Hypothèse de stabilité du prix du cuivre
- Pas d’intégration des aspects environnementaux (émissions, carburant)
- Complexité accrue si le nombre de variables augmente

---

### 💡 Ce que je peux en tirer pour mon projet

**Insights pratiques** :
1. L’approche hybride simulation + méta-modélisation permet d’explorer rapidement des scénarios et d’optimiser sous incertitude.
2. La méthode ε-contrainte est efficace pour gérer des objectifs multiples (production, coût).
3. Le recours à un surrogate model réduit fortement le coût de calcul par rapport à une optimisation directe par simulation.
4. Ce cadre est généralisable à d’autres mines ou systèmes industriels complexes.

- Titre : *A simulation model to study truck-allocation options*
- Fichier : `A simulation model to study truckallocation options.pdf`

---

## 📊 Résultats Clés

- Métriques utilisées :

    - Production totale (tonnage)
    - Temps d’attente aux pelles et dumps
    - Utilisation de la flotte
    - Congestion (bunching)
    - Effet des points de décision/intersections

- Chiffres importants :

    - La production augmente avec la taille de la flotte, puis se stabilise
    - Les stratégies dynamiques (MTWT, MSPR, GA, FDA) réduisent le bunching et améliorent l’utilisation de la flotte par rapport à l’affectation fixe (FTA)

    - Hypothèse d’absence de dépassement entre camions

- Hypothèses simplificatrices :

    - Simulation d’un quart de 8 heures
    - 100 réplications par scénario
    - Pas d’intégration des retards opérationnels réels

## 💡 Ce que je peux en tirer pour mon projet

    1. L’allocation dynamique améliore la performance globale du système camion-pelle.

    3. L’optimisation du semi-cycle (GA/FDA) offre un bon compromis entre production et congestion.
    4. La modélisation réaliste des interactions trafic-camions est essentielle pour la planification minière moderne.

## 💡 Synthèse des Limites Communes

Après lecture de ces études de cas, identifier les limites récurrentes des approches classiques :

### ⚠️ Problèmes Identifiés

1. **Incapacité à gérer l'incertitude dynamique** :
   - Articles : Upadhyay & Askari-Nasab (2017), Fioroni et al. (2008), Ozdemir (2019)
   - **→ Opportunité RL** : L'agent peut s'adapter en temps réel

2. **Optimisation locale vs globale** :
   - Articles : Ozdemir (2019), Çetin (2004)
   - **→ Opportunité RL** : Optimisation holistique de la flotte

3. **Hypothèses simplificatrices (routes parfaites, conditions stables)** :
   - Articles : Abolghasemian (2020), Çetin (2004), Zeng et al. (2022)
   - **→ Opportunité RL** : Simulation réaliste avec dégradation, météo, trafic

4. **Temps de calcul élevé (pour optimisation exacte)** :
   - Articles : Fioroni et al. (2008), Upadhyay & Askari-Nasab (2017)
   - **→ Opportunité RL** : Inférence rapide après entraînement

5. **Manque d'apprentissage continu** :
   - Articles : Choudhury & Naik, Zeng et al. (2022)
   - **→ Opportunité RL** : Amélioration continue avec nouvelles données

## 📄 Article 1 : Henderson et al. (2018) ⭐⭐⭐

**Référence complète** :
- Auteurs : Henderson, P. ; Islam, R. ; Bachman, P. ; Pineau, J. ; Precup, D. ; Meger, D.
- Année : 2018
- Titre : *Deep Reinforcement Learning That Matters*
- Source : Proceedings of the 32nd AAAI Conference on Artificial Intelligence, 3207-3214 — arXiv:1709.06560
- Fichier PDF : `Deep Reinforcement Learning That Matters.pdf` (présent)

**⭐ Intérêt** : Justifie plusieurs graines d'entraînement et la déclaration de tous les essais.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Les résultats de RL profond se reproduisent mal, et les améliorations annoncées sont difficiles à juger.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Expériences sur MuJoCo (TRPO, PPO, DDPG, ACKTR) en faisant varier hyperparamètres, architecture, échelle de la récompense, graines aléatoires, environnements et implémentations ; tests de significativité et intervalles par bootstrap.

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

- Forte sensibilité à tous ces facteurs. Avec les mêmes hyperparamètres, deux groupes de **5 graines** de TRPO sur HalfCheetah donnent des courbes de distributions significativement différentes.
- Deux implémentations d'un même algorithme ne sont pas équivalentes.
- Recommandations : rapporter graines et hyperparamètres, faire de nombreux essais, utiliser des tests et des intervalles de confiance.

---

## 📄 Article 2 : Agarwal et al. (2021) ⭐⭐⭐

**Référence complète** :
- Auteurs : Agarwal, R. ; Schwarzer, M. ; Castro, P. S. ; Courville, A. ; Bellemare, M. G.
- Année : 2021
- Titre : *Deep Reinforcement Learning at the Edge of the Statistical Precipice*
- Source : Advances in Neural Information Processing Systems 34, 29304-29320 — arXiv:2108.13264
- Fichier PDF : `Deep Reinforcement Learning at the Edge of the Statistical Precipice.pdf` (présent)

**⭐ Intérêt** : Source directe de notre agrégation sur les graines d'entraînement (§6.5).

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Les résultats sont comparés par des estimations ponctuelles (moyenne, médiane) obtenues avec très peu d'entraînements, sans tenir compte de l'incertitude.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Recommandations : intervalles de confiance par **bootstrap stratifié** ; **profils de performance** ; agrégats robustes : **moyenne interquartile** (IQM, moyenne des 50 % centraux), écart à l'optimum ; **probabilité d'amélioration** P(X > Y). Bibliothèque associée : rliable.

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Sur Atari 100k, les conclusions tirées des estimations ponctuelles changent avec une analyse plus complète ; mêmes incohérences sur ALE, Procgen et DeepMind Control.

---

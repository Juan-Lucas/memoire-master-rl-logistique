# Données d'équipement et de consommation

Données du constructeur (Caterpillar) et mesures de terrain sur le Cat 785D, camion de référence du simulateur. Elles fondent le modèle de carburant revu le 22/10/2026 (`simulation/constants.py`). Fiches créées le 22/10/2026.

**Référence complète** :
- Auteur : Caterpillar Inc.
- Année : 2019 (édition 49, SEBD0351-49, © 1979-2019 ; fichier nommé « 2020 »)
- Titre : *Caterpillar Performance Handbook*, Edition 49
- Source : Caterpillar, Peoria (IL), 2 448 pages
- Fichier PDF : `Cat-Performance-Handbook-2020.pdf` (présent)

**⭐ Intérêt** : Source publiée de la consommation horaire du Cat 785D, qui fixe le niveau moyen de notre modèle de carburant.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Manuel de référence du constructeur : caractéristiques, performances (effort à la jante, vitesse, pente, freinage) et coûts de possession et d'exploitation de tous les engins Caterpillar.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Tableaux par famille d'engins. Pour les camions miniers, la section 25 « Owning & Operating Costs » donne des **tables de consommation horaire** par usage léger, moyen et intensif, établies à partir des machines qui remontent leurs données par la télématique VisionLink (note 2 du tableau).

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

Valeurs vérifiées pour notre camion de référence :
- **Cat 785D (1 450 hp)** : 54,2-81,4 L/h (léger), **81,4-108,5 L/h (moyen)**, 108,5-135,6 L/h (intensif) ; 785C : 53,7-80,6 / 80,6-107,5 / 107,5-134,4 L/h.
- **Spécifications du 785D** (section 10, « Mining & Off-Highway Trucks ») : charge cible **133,0 t**, poids à vide en ordre de marche 116 525 kg, poids total cible 249 476 kg, benne 78 m³ (2:1 SAE).
- Également disponibles et non exploités : courbes effort à la jante, vitesse et pente, et performances de freinage du 785D (section 10), qui pourraient servir à vérifier les vitesses du simulateur.

---

### 🔗 Pour mon article

- Le simulateur vise **95 L/h** (milieu de la plage « usage moyen ») pour l'Affectation fixe en nominal ; la borne de normalisation de l'observation est le maximum publié (135,6 L/h × 8 h).
- ⚠️ Le code citait « édition 2022 » : corrigé en **édition 49 (2019)**, celle de ce fichier.
- `scenarios.py` parle de Cat 785C et `constants.py` de 785D : les consommations sont quasi identiques, harmoniser le texte sur le 785D.
- À ajouter au `.bib` (`@manual`).

## 📄 Article 3 : Kubler (2015) ⭐⭐⭐

**Référence complète** :
- Auteur : Kubler, K. A. C.
- Année : 2015
- Titre : *Optimisation of off-highway truck fuel consumption through mine haul road design*
- Source : Mémoire de Bachelor of Engineering (Mechanical), University of Southern Queensland, Faculty of Health, Engineering and Sciences, octobre 2015, 145 p.
- Fichier PDF : `Kubler_K_Baker.pdf` (présent)

**⭐ Intérêt** : Seule source de **mesures** de consommation d'un Cat 785D dont nous disposons ; elle fonde le ralenti et l'effet de la pente de notre modèle de carburant.

> Fiche du 22/10/2026, fiche rédigée à partir du **texte intégral**. Lecture assistée : à relire.

---

### ❓ Question 1 : Quel est le problème principal ?

**Réponse** :

Optimiser la consommation des camions Cat 785D d'une mine de charbon à ciel ouvert du Queensland par la conception des pistes : quantifier la consommation en rampe, en arrêt-redémarrage et au ralenti.

---

### ❓ Question 2 : Comment les auteurs s'y prennent-ils ?

**Réponse** :

Données de surveillance de l'état d'un camion, combinées : Caterpillar VIMS (moteur), débitmètres de carburant, GPS SAFEmine. Sélection des échantillons selon la position de l'accélérateur et la charge du moteur (exclusion des roues libres et freinages) ; ajustement de polynômes ; **un seul camion** étudié (limite déclarée par l'auteure).

---

### ❓ Question 3 : Quelle est la conclusion principale ?

**Réponse** :

- **Ralenti : 11,3 L/h** en moyenne (écart-type 1,71 ; de 8 à 14 L/h ; 74 095 échantillons d'une seconde ; 700 tr/min, charge moteur 3,6 %) (§5.3, p. 67).
- **Rampes** (§5.1.3, p. 57), en L/km selon la pente x (fraction) : à vide f = −1661 x³ + 727,7 x² + 6,809 x + 4,562 ; chargé f = −803,8 x³ + 603,87 x² + 80,825 x + 7,638 (domaine observé : jusqu'à environ 18 % à vide et 23 % en charge). Soit environ 4,7 et 8,5 L/km à 1 %, 11 et 21 L/km à 10 %.
- **Arrêt-redémarrage** (§5.2) : environ 310 mL par arrêt pour un camion chargé, 195 mL à vide.
- Résultats cohérents avec les moyennes horaires du Handbook et les données de ravitaillement du site.

---

### 🔗 Pour mon article

- **Modèle de carburant du simulateur (22/10/2026)** : ralenti 11,3 L/h (au lieu de 64,9) ; roulage selon ses courbes de pente, multipliées par 0,4466 pour retrouver 95 L/h (Handbook). Sans ce facteur, le simulateur consommerait 211 L/h : toutes ses pistes montent, y compris au retour.
- Répond à la note M20 de l'article (« confronter le modèle de carburant à une donnée de constructeur ou de terrain »).
- Mémoire de licence, non évalué par les pairs : à citer comme source de données de terrain, avec cette réserve et la limite « un seul camion ».
- À ajouter au `.bib` (`@thesis` ou `@misc`).
